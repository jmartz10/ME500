```python
import numpy as np
import matplotlib.pyplot as plt

plt.style.use('seaborn-v0_8-whitegrid')
```

```python
def newton_linear(t_data, y_data, t):
    t0, t1 = t_data 
    y0, y1 = y_data

    return y0 + ((y1-y0)/(t1 - t0))*(t - t0)
```

```python
# Problem 1(a)
t_data = np.array([1, 6])
y_data = np.array([0, 1.792])
t_target = 2.0

p1_newton = newton_linear(t_data[:2], y_data[:2], t_target)

print(f"Newton Linear: {p1_newton:.5f}")
```

```python
def newton_quadratic(t_data, y_data, t):
    t1, t2, t3, = t_data
    y1, y2, y3 = y_data

    f_t1_t2 = (y2 - y1)/(t2 - t1)
    f_t2_t3 = (y3 - y2)/(t3 - t2)
    f_t1_t2_t3 = (f_t2_t3 - f_t1_t2)/(t3 - t1)

    return y1 + f_t1_t2*(t - t1) + f_t1_t2_t3 * (t - t1)*(t - t2)
```

```python
# Problem 1(b)
t_data = np.array([1, 4, 6])
y_data = np.array([0, 1.386, 1.792])
t_target = 2.0

p2_newton = newton_quadratic(t_data, y_data, t_target) 

print(f"Exact ln(2): {np.log(2):.5f}")
print(f"Newton Quadratic: {p2_newton:.5f}")
```

```python
def newton_interpolation(t_data, y_data, t, n):

    t_n = t_data[:n+1]
    y_n = y_data[:n+1]

    A = np.ones((n+1, n+1))

    for i in range(n+1):
        for j in range(1, n+1):
            A[i, j] = A[i, j-1] * (t_n[i] - t_n[j-1])
    x = np.linalg.solve(A, y_n)

    value = x[0]
    product = 1.0

    for j in range(1, n+1):
        product = product * (t - t_n[j-1])
        value = value + x[j] * product
    return value 
```

```python
# Problem 1(c)
t_data = np.array([1, 4, 6, 8])
y_data = np.log(t_data)
t_target = 2.0
for n in range(1, 4):
    p = newton_interpolation(t_data, y_data, t_target, n)
    print(f"n = {n}: {p:.5f}")
print(f"Exact ln(2): {np.log(2):.5f}")
```

```python
def lagrange_linear(t_data, y_data, t):
    t1, t2, = t_data
    y1, y2 = y_data

    L1 = (t - t2)/(t1 - t2)
    L2 = (t - t1)/(t2 - t1)

    return y1*L1 + y2*L2
```

```python
# Problem 2(a)
t_data = np.array([1, 6])
y_data = np.array([0, 1.792])
t_target = 2.0

p1_lagrange = lagrange_linear(t_data[:2], y_data[:2], t_target)

print(f"Lagrange Linear: {p1_lagrange:.5f}")
```

```python
def lagrange_quadratic(t_data, y_data, t):

    t1, t2, t3 = t_data
    y1, y2, y3 = y_data

    L1 = (t - t2) * (t - t3) / ((t1 - t2) * (t1 - t3))
    L2 = (t - t1) * (t - t3) / ((t2 - t1) * (t2 - t3))
    L3 = (t - t1) * (t - t2) / ((t3 - t1) * (t3 - t2))

    return y1*L1 + y2*L2 + y3*L3 
```

```python
# Problem 2(b)
t_data = np.array([1, 4, 6])
y_data = np.array([0, 1.386, 1.792])
t_target = 2.0

p2_lagrange = lagrange_quadratic(t_data, y_data, t_target) 

print(f"Exact ln(2): {np.log(2):.5f}")
print(f"Lagrange Quadratic: {p2_lagrange:.5f}")
```

```python
def lagrange_interpolation(t_data, y_data, t, n):
    t_n = t_data[:n+1]
    y_n = y_data[:n+1]
    m = len(t_n)

    t_eval_arr = np.atleast_1d(t)
    results = []

    for t_eval in t_eval_arr:
        val = 0.0

        for i in range(m):
            L = 1.0

            for j in range(m):
                if j !=i:
                    L = L * (t_eval - t_n[j])/(t_n[i] - t_n[j])
                    
            val = val + y_n[i]*L

        results.append(val)

    return results[0] if len(results) == 1 else np.array(results)
```

```python
# Problem 2(c)
t_data = np.array([1, 4, 6, 8])
y_data = np.log(t_data)
t_target = 2.0
for n in range(1, 4):
    p = lagrange_interpolation(t_data, y_data, t_target, n)
    print(f"n = {n}: {p:.5f}")
print(f"Exact ln(2): {np.log(2):.5f}")
```
