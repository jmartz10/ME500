```python
import numpy as np
import matplotlib.pyplot as plt
import math

np.set_printoptions(precision=4, suppress=True)
```

```python
#Problem 1(a)
def newton_cotes(f, a, b):
    
    M = (b-a)*f((a+b)/2)

    
    T = (b-a)/2 * (f(a) + f(b))

    S = (b-a)/6 * (f(a) + 4*f((a+b)/2) + f(b))

    return {'midpoint': M, 'trapezoidal': T, 'simpsons': S}

#Problem 1(b)
f1 = lambda x: x**2 + 1
a1, b1 = 0, 2
exact1 = 14 / 3 


results = newton_cotes(f1, a1, b1)


print(f"Exact: {exact1:.4f}")
for method, val in results.items():
    error = abs(val - exact1)
    print(f"{method.capitalize():<13}: {val:.4f} | Error: {error:.4f}")
```

```python
#Problem 2(a)
def gauss_legendre_2(f, a, b):
    
    xd = np.array([-1/np.sqrt(3), 1/np.sqrt(3)])
    w = np.array([1.0, 1.0])

    
    x = ((b-a)/2 * xd) + ((a+b)/2)

   
    integral = (b-a)/2 * np.sum(w * f(x))

    return integral

#Problem 2(b)
f2 = lambda x: x**3 + 2*x
a2, b2 = 0, 0.8
exact2 = 0.7424  


gl_approx = gauss_legendre_2(f2, a2, b2)
error2 = abs(gl_approx - exact2)

print(f"Exact integral:       {exact2:.4f}")
print(f"GL2 approximation:    {gl_approx:.4f}")
print(f"Absolute error:       {error2:.4f}")
```

```python
#Problem 3(a)
def get_diff_weights(delta, max_deriv):
   
    n = len(delta)
    j = np.arange(n)

    
    j_fact = np.array([math.factorial(k) for k in j])

    V = (delta[:, None] ** j)/j_fact

    
    C = np.linalg.inv(V)

    return C[:max_deriv+1, :]


delta = np.array([0, 1, 2, 3])
C = get_diff_weights(delta, 3)
print("Differentiation Weight Matrix C (up to 3rd derivative):")
print(C)
```

```python
#Problem 3(b)
x_eval = 1.0
h = 0.1

f_vals = np.sin(x_eval + delta*h)

df_approx = (1/h)*np.sum(C[1, :] * f_vals)

df_exact = np.cos(x_eval)

print(f"Approximate f'({x_eval}): {df_approx:.6f}")
print(f"Exact f'({x_eval}):       {df_exact:.6f}")
print(f"Absolute error:           {abs(df_approx - df_exact):.6f}")
```
