```python
import numpy as np
import matplotlib.pyplot as plt

# Set a clean plot style
plt.style.use('seaborn-v0_8-whitegrid')
```

```python
# Problem 1(a)
N = 4000
x = np.linspace(-10*np.pi, 10*np.pi, N)
y = np.zeros(N)

i = 0
while i<N:
    if x[i] < 0:
        y[i] = np.sin(x[i]) / x[i]
    else:
        y[i] = 1

    i = i + 1

plt.plot(x, y, color='blue')
plt.title("Jordan's Plot")
plt.xlabel("x")
plt.ylabel("$f(x)$")
plt.show()
```

```python
# Problem 1(b)
y2 = np.zeros(N)

for i in range(N):
    if x[i] < 0:
        y2[i] = np.sin(x[i]) / x[i]
    else:
        y2[i] = 1

plt.plot(x, y2, color='purple')
plt.title("Jordan's 2nd Plot")
plt.xlabel("x")
plt.ylabel("$f(x)$")
plt.show()
```

```python
# Problem 2 Sum Calculations
def calc_sum(n):
    s = 0
    for i in range(1, n+1):
        s = s + i**2
    return s

n_vals = np.array([10, 100, 1000])
s_vals = np.array([calc_sum(n) for n in n_vals])

plt.plot(n_vals, s_vals, color='turquoise', marker='o')
plt.title("Jordan's 3rd Plot")
plt.xlabel("n")
plt.ylabel("$s(n)$")
plt.show()
```

```python
# Problem 3(a)
def exact(n):
    return n*(n+1)*(2*n+1)/6

def approx(n):
    return n**3/3
```

```python
# Problem 3(b)
def error_abs(n):
    return np.abs(approx(n) - exact(n))

def error_rel(n):
    return error_abs(n) / np.abs(exact(n))
```

```python
# Problem 3(c)
n_vals = np.array([10, 100, 1000])

print("Absolute errors:")
print(error_abs(n_vals))

print("Relative errors:")
print(error_rel(n_vals))
```

```python
n_plot = np.arange(1, 1001)
plt.loglog(n_plot, error_abs(n_plot))
plt.title("Jordan's Absolute Error Plot")
plt.xlabel("n")
plt.ylabel("Absolute Error")
plt.show()
```

```python
plt.loglog(n_plot, error_rel(n_plot))
plt.title("Jordan's Relative Error Plot")
plt.xlabel("n")
plt.ylabel("Relative Error")
plt.show()
```
